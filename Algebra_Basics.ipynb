#### Set up a proportion:
$$\frac{n1}{d1} = \frac{n2}{d2}$$


Screenshot 2026-07-25 162505.jpg


```python
# Put a zero in for the unknown value
n1 = 1
d1 = 3
n2 = 3
d2 = 0

ans = 0

if n2 == 0:
  ans = d2 * n1 / d1
  print('n2 = ', ans)
elif d2 == 0:
  ans = n2 * d1 / n1
  print('d2 = ', ans)
```

#### Solving for x
The variable `eq` is the equation *set* **equal to 0** will return an array of solutions

```python
from math import e
import sympy
from sympy import symbols
from sympy.solvers import solve

x = symbols('x')

# put the equation here
eq = 3 * x - 2

print(f'x = {solve(eq, x)}')
eq = 3**x - 2

print(f'x = {solve(eq, x)}')
eq = x**2 - 4

print(f'x = {solve(eq, x)}')
eq = x**2 + 4
print(f'x = {solve(eq, x)}')
eq = x**2 - 2

print(f'x = {solve(eq, x)}')
eq = x**2 + 2

print(f'x = {solve(eq, x)}')
```

```python
# 1. Get the list of all solutions
m = solve(eq, x)

# 2. Print the first solution safely
if m:
    print(f'x = {m}')

# 3. Correctly loop through every solution in the list
for solution in m:
    print(f'solutions of the eqn are: {solution}')
```
